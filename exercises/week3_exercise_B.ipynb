### Sentinel-1 メタデータ例（参考）

実際の Sentinel-1 IW シーンのメタデータには以下の情報が含まれています：

| 項目 | 典型的な値 | 注意点 |
|------|-----------|-------|
| instrumentMode | IW (Interferometric Wide swath) | 農地解析に最適（10m x 10m） |
| orbitProperties_pass | ASCENDING / DESCENDING | 統一しないと入射角が変わる |
| relativeOrbitNumber_start | ex. 107 | 同じ番号＝同じ入射角 |
| transmitterReceiverPolarisation | VV, VH | 両偏波同時取得が標準 |
| incidenceAngle | 29.1°〜46.0° | 近端＜遠端で約17°の幅 |
| resolution | 20m (range) x 22m (azimuth) | マルチルック後の製品分解能 |
| swath | IW1 / IW2 / IW3 | 3つのサブスワスで構成 |
